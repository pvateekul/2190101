```python
# Solution!
# Create a function to count word frequency using a dictionary
def count_words(text):
    words = text.split()
    freq = {}
    for word in words:
        if word in freq:
            freq[word] += 1
        else:
            freq[word] = 1
    return freq

count_words("This is a test. This is only a test.")
```

```python
# Solution!
def count_word_frequency(text):
    # Convert text to lowercase and split by whitespace
    words = text.lower().split()

    # Initialize an empty dictionary to hold word counts
    frequency = {}

    # Count the occurrences of each word
    for word in words:
        # Clean basic punctuation from the ends of words
        cleaned_word = word.strip('.,!?;:"()')
        if cleaned_word:
            frequency[cleaned_word] = frequency.get(cleaned_word, 0) + 1

    return frequency

# Example usage:
sample_text = "This is a test. This is only a test."
result = count_word_frequency(sample_text)
print(result)
```

## <b>5) Transform Code (Refactoring / Alternative Syntax)</b>

Code transformation involves taking working code and changing its structure, style, or paradigm (for example, making it cleaner, converting languages, or optimizing performance) without changing what it actually does.

### **Example: Transforming a `for` loop into a List Comprehension**
We will take a classic multi-line loop that filters and squares numbers and transform it into a single-line Python list comprehension.

```python
# --- Original Code (Traditional For-Loop) ---
numbers = [1, 2, 3, 4, 5, 6]
squared_evens = []
for n in numbers:
    if n % 2 == 0:
        squared_evens.append(n ** 2)

print("Original Loop Result:", squared_evens)

# --- Transformed Code (Concise List Comprehension) ---
transformed_squared_evens = [n ** 2 for n in numbers if n % 2 == 0]

print("Transformed Code Result:", transformed_squared_evens)
```
